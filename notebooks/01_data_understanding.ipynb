### Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
import seaborn as sns

from pathlib import Path
```

### Loading Dataset

```python
path = Path("../data/raw/stores_sales.csv")
df = pd.read_csv(path)
df.head(1)
```

### EDA

```python
df.info()
df.describe()
df.isna().sum()
```

```python
sns.heatmap(df.corr(numeric_only=True), cmap="coolwarm", annot=True)
df[["Sales", "Profit"]].hist(bins=50, figsize=(8,4))
```

Sales (vänster): kraftigt höger-skev. De flesta ordrar har låg försäljning, men några få har extremt hög. Stor spridning och ser ut att finnas tydliga outliers.

Profit (höger): centrerad kring 0, med många små vinster och några förluster. Vinsten tycks varierar mycket mellan produkter.

Sales ej normalfördelad. Profit tycks vara nära normalfördelad även om väldigt spetsig.
