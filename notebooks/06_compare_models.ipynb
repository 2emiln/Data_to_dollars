### Imports

```python
import pandas as pd
import numpy as np
import joblib
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
X_test = pd.read_csv("../data/processed/X_test.csv")
y_test_sales  = pd.read_csv("../data/processed/y_test_sales.csv").squeeze("columns")
y_test_profit = pd.read_csv("../data/processed/y_test_profit.csv").squeeze("columns")
```

```python
models = [
    ("sales_lr", "../models/sales_lr.pkl", y_test_sales),
    ("sales_rf", "../models/sales_rf.pkl", y_test_sales),
    ("sales_xgb_best", "../models/sales_xgb_best.pkl", y_test_sales),
    #("profit_lr", "../models/profit_lr.pkl", y_test_profit),
    ("profit_rf", "../models/profit_rf.pkl", y_test_profit),
    ("profit_xgb_best", "../models/profit_xgb_best.pkl", y_test_profit),
]
```

```python
results = []
for name, path, y_test in models:
    model = joblib.load(path)
    pred = model.predict(X_test)
    mae  = mean_absolute_error(y_test, pred)
    rmse = np.sqrt(mean_squared_error(y_test, pred))
    r2   = r2_score(y_test, pred)
    results.append({"Model": name, "MAE": mae, "RMSE": rmse, "R2": r2})

df = pd.DataFrame(results)
df.to_csv("../reports/ml_models_metrics.csv", index=False)
df
```
