### Imports

```python
import pandas as pd
import numpy as np
import joblib

from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from pathlib import Path

RANDOM_STATE = 42
```

### Loading

```python
path = Path("../models/preprocessor.pkl")
pre = joblib.load(path)
X_train = pd.read_csv("../data/processed/X_train.csv")
X_test = pd.read_csv("../data/processed/X_test.csv")
y_train_sales = pd.read_csv("../data/processed/y_train_sales.csv").squeeze("columns")
y_test_sales = pd.read_csv("../data/processed/y_test_sales.csv").squeeze("columns")
y_train_profit = pd.read_csv("../data/processed/y_train_profit.csv").squeeze("columns")
y_test_profit = pd.read_csv("../data/processed/y_test_profit.csv").squeeze("columns")
```

### Baseline model

```python
model = Pipeline([("pre", pre), ("est", LinearRegression())])
model.fit(X_train, y_train_sales)
pred = model.predict(X_test)

mae = mean_absolute_error(y_test_sales, pred)
mse = mean_squared_error(y_test_sales, pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test_sales, pred)
```

```python
print({
    "MAE": mae,
    "MSE": mse,
    "RMSE": rmse,
    "R2": r2
})
```

```python
joblib.dump(model, "../models/sales_lr.pkl")
```
