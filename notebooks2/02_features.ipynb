```python
import pandas as pd
from datetime import date as dt
```

```python
df = pd.read_csv("../data/processed/stores_sales_cleaned.csv", parse_dates=["Order Date", "Ship Date"])
```

### Overview

```python
df.info()
```

```python
print(df[["Order Date","Ship Date"]].min())
print(df[["Order Date","Ship Date"]].max())
```

```python
# Skapar nya features

df["Order Year"] = df["Order Date"].dt.year
df["Order Quarter"] = df["Order Date"].dt.quarter
df["Order Month"] = df["Order Date"].dt.month
df["Order Day"] = df["Order Date"].dt.day
df["Order Weekday"] = df["Order Date"].dt.weekday
df["Order Weekend"] = (df["Order Date"].dt.weekday >=5)

```
